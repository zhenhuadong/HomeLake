```python
import pandas as pd
import numpy as np

# 设置随机种子，保证结果可复现
np.random.seed(42)

# ==================== 1. 生成学生表 ====================
num_students = 20  # 学生人数

# 学号（例如：20240001 ~ 20240020）
student_ids = [f'2024{i:04d}' for i in range(1, num_students + 1)]

# 随机生成姓名（百家姓 + 常见名字）
surnames = ['张', '王', '李', '刘', '陈', '杨', '赵', '黄', '周', '吴']
given_names = ['明', '伟', '芳', '强', '丽', '静', '军', '勇', '敏', '杰']
names = [np.random.choice(surnames) + np.random.choice(given_names) for _ in range(num_students)]

# 性别（随机生成男女）
genders = np.random.choice(['男', '女'], size=num_students, p=[0.5, 0.5])

# 班级（例如 1~4 班）
classes = [f'{np.random.randint(1, 5)}班' for _ in range(num_students)]

# 创建学生表 DataFrame
students_df = pd.DataFrame({
    '学号': student_ids,
    '姓名': names,
    '性别': genders,
    '班级': classes
})

# ==================== 2. 生成成绩表 ====================
subjects = ['语文', '数学', '英语', '物理', '化学']  # 科目列表
num_subjects = len(subjects)

# 为每个学生在每个科目下生成一条成绩记录
score_records = []
for sid in student_ids:
    for subject in subjects:
        # 成绩范围 60~100 分，部分不及格（40~59）也可模拟真实情况
        # 这里采用正态分布 + 截断，使大部分集中在 70~90 之间
        score = int(np.random.normal(75, 12))
        score = max(40, min(100, score))  # 限制在 [40, 100]
        score_records.append({
            '学号': sid,
            '科目': subject,
            '成绩': score
        })

scores_df = pd.DataFrame(score_records)

# ==================== 3. 查看数据示例 ====================
print("学生表（前5行）：")
print(students_df.head())
print("\n学生表基本信息：")
print(students_df.info())

print("\n成绩表（前5行）：")
print(scores_df.head())
print("\n成绩表统计描述：")
print(scores_df['成绩'].describe())

# ==================== 4. 可选：保存为 CSV 文件 ====================
students_df.to_csv('students.csv', index=False, encoding='utf-8-sig')
scores_df.to_csv('scores.csv', index=False, encoding='utf-8-sig')
```

```python
students_df=pd.read_csv('students.csv')
scores_df=pd.read_csv('scores.csv')
```

```python
df = students_df.merge(scores_df, left_on="学号", right_on="学号", how="inner")
# 每个班级的平均成绩
df.groupby("班级")["成绩"].mean().sort_values(ascending=False)
# 每个科目的平均成绩
df.groupby("科目")["成绩"].mean().sort_values(ascending=False)
# 每个性别的平均成绩
df.groupby("性别")["成绩"].mean().sort_values(ascending=False)
# 每个学生的平均成绩
df.groupby("姓名")["成绩"].mean().sort_values(ascending=False).head(10)
# 每个班级和科目的最高成绩
df2=df.loc[df.groupby(["班级", "科目"])["成绩"].idxmax()]

# 每个班级和科目的成绩透视表 科目为行，班级为列，最高成绩为值
df2[["班级", "科目", "成绩"]].pivot(index="科目", columns="班级", values="成绩")
df2[["班级", "科目", "成绩"]].pivot_table(index="科目", columns="班级", values="成绩", aggfunc="max")
# 每个班级和科目的成绩透视表 科目为行，班级为列，平均成绩为值
df.pivot_table(index="科目", columns="班级", values="成绩", aggfunc="mean")

```
